```python
import pandas as pd

DATA_PATH = "../data/ict3113_tickets.csv"
SHUFFLED_PATH = "../data/shuffled_golden_test.csv"

df = pd.read_csv(DATA_PATH)
print(df.shape)
```

```python
# Keep only rows 5000-5999 (filtering on the `row` column so re-running is safe)
df = df[df["row"].between(5000, 5999)]
shuffled_df = df.sample(frac=1, random_state=42).reset_index(drop=True)
shuffled_df = shuffled_df[:150]
shuffled_df.to_csv(SHUFFLED_PATH, index=False)
print(shuffled_df.shape)
```
